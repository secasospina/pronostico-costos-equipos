Pruebas de conexión con Gemini y de las herramientas del agente que presenta los resultados del análisis.

```python
import os
from dotenv import load_dotenv
from google import genai

load_dotenv("../.env")
cliente = genai.Client(api_key=os.getenv("GEMINI_API_KEY"))

for modelo in cliente.models.list():
    if "generateContent" in (modelo.supported_actions or []):
        print(modelo.name)
```

```python
MODELO = "gemini-3.5-flash-lite"

respuesta = cliente.models.generate_content(
    model=MODELO,
    contents="En una frase, ¿qué es un agente de inteligencia artificial?"
)
print(respuesta.text)
```
