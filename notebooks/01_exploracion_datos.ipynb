# 01 - Exploración y limpieza de datos
Carga de las series de materias primas (X, Y, Z) y del histórico de equipos para revisar su estructura.

```python
from pathlib import Path
import pandas as pd

DATA_RAW = Path("../data/raw")
DATA_PROCESSED = Path("../data/processed")
```

```python
x = pd.read_csv(DATA_RAW / "X.csv")
display(x.head())
x.info()
```

```python
y = pd.read_csv(DATA_RAW / "Y.csv", sep=";", decimal=",", encoding="utf-8-sig")
display(y.head())
y.info()
```

```python
z = pd.read_csv(DATA_RAW / "Z.csv")
display(z.head())
z.info()
```

```python
historico_equipos = pd.read_csv(DATA_RAW / "historico_equipos.csv")
display(historico_equipos.head())
historico_equipos.info()
```

```python
y.columns.tolist()
```

```python
x.columns.tolist()
```

```python
x["Date"] = pd.to_datetime(x["Date"], format="%Y-%m-%d")
x.info()
```

```python
z["Date"] = pd.to_datetime(z["Date"], format="%Y-%m-%d")
z.info()
```

```python
historico_equipos["Date"] = pd.to_datetime(historico_equipos["Date"], format="%Y-%m-%d")
historico_equipos.info()
```

```python
y["Date"] = pd.to_datetime(y["Date"], format="%d/%m/%Y")
y.info()
```

```python
print("X:", x["Date"].min(), "a", x["Date"].max())
print("Duplicados:", x["Date"].duplicated().sum())
```

```python
print("Y:", y["Date"].min(), "a", y["Date"].max())
print("Duplicados:", y["Date"].duplicated().sum())
```

```python
print("Z:", z["Date"].min(), "a", z["Date"].max())
print("Duplicados:", z["Date"].duplicated().sum())
```

```python
print("historico_equipos:", historico_equipos["Date"].min(), "a", historico_equipos["Date"].max())
print("Duplicados:", historico_equipos["Date"].duplicated().sum())
```

```python
x = x.sort_values("Date").reset_index(drop=True)
display(x.head())
```

```python
y = y.sort_values("Date").reset_index(drop=True)
display(y.head())
```

```python
z = z.sort_values("Date").reset_index(drop=True)
display(z.head())
```

```python
historico_equipos = historico_equipos.sort_values("Date").reset_index(drop=True)
display(historico_equipos.head())
```

```python
comparacion_x = pd.merge(
    historico_equipos[["Date", "Price_X"]],
    x,
    on="Date",
    how="inner"
)
display(comparacion_x.head())
```

```python
print("Diferencias en X:", (comparacion_x["Price_X"] != comparacion_x["Price"]).sum())
```

```python
comparacion_y = pd.merge(
    historico_equipos[["Date", "Price_Y"]],
    y,
    on="Date",
    how="inner"
)
display(comparacion_y.head())
```

```python
print("Diferencias en Y:", (comparacion_y["Price_Y"] != comparacion_y["Price"]).sum())
```

```python
comparacion_z = pd.merge(
    historico_equipos[["Date", "Price_Z"]],
    z,
    on="Date",
    how="inner"
)
display(comparacion_z.head())
```

```python
print("Diferencias en Z:", (comparacion_z["Price_Z"] != comparacion_z["Price"]).sum())
```

### Consistencia entre el histórico y las series originales
Los precios de X, Y y Z del histórico coinciden al 100% con los archivos originales en las 3.530 fechas. El histórico es una base confiable para el análisis.

```python
saltos = historico_equipos["Date"].diff().dt.days
display(saltos.value_counts().sort_index())
```

### Días faltantes en el histórico
- 2.808 saltos de 1 día los cuales corresponden a días hábiles consecutivos.
- 687 saltos de 3 días que corresponden a los  fines de semana.
- 9 saltos de 2 días y 25 de 4 días: festivos entre semana y puentes.

**Conclusión:** el salto máximo es de 4 días, así que no hay huecos grandes en la información. Los datos corresponden a días hábiles y no requieren imputación.
